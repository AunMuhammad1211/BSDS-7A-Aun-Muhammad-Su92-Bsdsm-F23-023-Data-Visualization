## Channel allocation

I assigned each variable to the channel that makes the required comparison as accurate as possible. GDP per capita is on the **x-position** with a logarithmic scale because the variable is multiplicative and contains a very large upper-tail outlier; position is the most precise channel for comparing numeric values. Healthcare access is on the **y-position**, bounded from 0 to 100, for the same reason. Population is encoded by **marker area**, not radius, because it is an amount and area is the visual quantity represented by a bubble. I used `s=(20, 900)` points?, so the diameter ratio is `sqrt(900/20) = 6.708`, and included a legend with real population values. Region is encoded by **categorical hue** because it identifies six groups rather than measuring an amount. I retained and directly labelled the maximum GDP observation instead of deleting it; the log axis makes its multiplicative separation visible without allowing it to determine the rest of the display.

## Parallel-coordinates decisions

The five axes use **min?max scaling** to put GDP, access, urban share, median age, and the already-normalized case rate into the common 0?1 frame. The scaling decision is printed on the figure because the axes are otherwise unitless. I computed the order using the greedy absolute-correlation chain, rather than choosing it by eye: **healthcare_access ? cases_per_100k ? urban_pct ? median_age ? gdp_pc**. The strongest adjacent relationship is healthcare access versus case rate, with `r = ?0.799`, producing the visible X-shaped waist. This view reveals crossing and shared trajectory patterns across all five variables at once: for example, lines that move from high access to low incidence can be followed into their urbanization and age positions. A pairwise scatter matrix would show individual pairs, but it would not show the same multi-variable paths and crossings in one ordered display. To control clutter, all countries are grey and one region is brushed in blue.

## Spatial decisions

The choropleth uses **cases per 100,000 residents**, not raw cases, because population is the exposure denominator and the question is incidence/risk rather than volume. I used five **quantile classes** with the breaks printed in the legend; this gives all five classes observations while the caption makes clear that the classes are comparative, not natural thresholds. The map uses a local **equal-area projection approximation** before drawing and reasoning about area, because colour fills occupy geographic areas and degree coordinates would distort area comparisons across latitude. The district join was checked: all 24 rows matched and zero failed. The caption identifies the 24 district boundaries, synthetic Week 5 source, extraction date, projection, and that no unit was suppressed or merged. A sorted raw-count bar chart is placed beside the map so volume remains available without confusing it with incidence.

## Counterfactual and ethics

If I had published the raw-count map alone, it would have pointed readers toward the most populous districts?**D17, D13, and D09**?because they have the largest case totals. That map would largely be a population map: cases and population correlate at `r = +0.986`. The normalized rate map instead identifies **D19, D20, and D21**, with zero overlap with the count top three. This is not a cosmetic distinction. Repeatedly colouring D17, D13, or D09 dark could attach an outbreak reputation to places partly because many people live there, while obscuring the higher per-person incidence in D19?D21. The Part C reader study should be used here as evidence about actual decisions: the saved reader results must contain at least five real readers before I report a mean overlap; I do not fabricate that result. No district was suppressed or merged in this work, so there is no small-count suppression rule or hidden count to disclose.


## Reproducible evidence

The next cell checks the numerical claims and reports the status of the real-reader evidence file. Run it after completing Part C.


```python
from pathlib import Path
import pandas as pd
from scipy import stats
BASE=Path.cwd()
if not (BASE/'district_health.csv').exists() and (BASE/'Week5').exists(): BASE=BASE/'Week5'
district=pd.read_csv(BASE/'district_health.csv')
district['rate']=district.cases/district.population*1e5
print('District join expectation: 24 rows; failures reported in Task4: 0')
print('Pooled rate:',round(district.cases.sum()/district.population.sum()*1e5,1),'per 100,000')
print('Top count districts:',district.nlargest(3,'cases').code.tolist())
print('Top rate districts:',district.nlargest(3,'rate').code.tolist())
print('Top-three overlap:',len(set(district.nlargest(3,'cases').code)&set(district.nlargest(3,'rate').code)))
print('r(cases,population):',round(stats.pearsonr(district.cases,district.population).statistic,3))
reader_file=BASE/'partC_reader_results.csv'
if not reader_file.exists():
    print('Reader evidence: file missing; collect five real readers before submission.')
else:
    readers=pd.read_csv(reader_file)
    if len(readers)<5:
        print(f'Reader evidence: {len(readers)} rows found; at least five real readers are required. Mean overlap is not reported.')
    else:
        print('Mean reader overlap:',round(readers.overlap_count.mean(),3))
```
